### Gradient Descent

````python
# Assuming df and f are defined functions for the derivative and the function itself

# Gradient Descent Parameters
new_x = 3
previous_x = 0
step_multiplier = 0.1
precision = 0.00001

x_list = [new_x]
slope_list = [df(new_x)]  # Append the initial slope

for n in range(500):
    previous_x = new_x
    gradient = df(previous_x)  # Calculate the gradient at the previous x
    new_x = previous_x - step_multiplier * gradient  # Update new_x using the gradient

    step_size = abs(new_x - previous_x)

    x_list.append(new_x)
    slope_list.append(df(new_x))  # Append the slope at the new_x

    if step_size < precision:
        print('Loop ran this many times:', n)
        break

print('Local minimum occurs at:', new_x)
print('Slope or df(x) value at this point is:', df(new_x))
print('f(x) value or cost at this point is:', f(new_x))
````

## Gradient Descent

````python
def gradient_descent(derivative_func, initial_guess, multiplier=0.02, precision=0.001,
                    max_iter=300):
    new_x = initial_guess
    x_list = [new_x]
    slope_list = [derivative_func(new_x)]

    for n in range(max_iter):
        previous_x = new_x
        gradient = derivative_func(previous_x)
        new_x = previous_x - multiplier * gradient

        step_size = abs(new_x - previous_x)
        x_list.append(new_x)
        slope_list.append(derivative_func(new_x))

        if step_size < precision:
            break
    return new_x, x_list, slope_list
````

````python
import numpy as np
import matplotlib.pyplot as plt

# Define the function g(x) = x^2 and its derivative dg(x)/dx = 2x
def g(x):
    return x**2

def dg(x):
    return 2*x

# Gradient Descent Function
def gradient_descent(derivative_func, initial_guess, multiplier, precision, max_iter):
    x = initial_guess
    x_list = [x]
    for _ in range(max_iter):
        gradient = derivative_func(x)
        x = x - multiplier * gradient  # Update x based on the gradient and multiplier
        x_list.append(x)
        if abs(x_list[-1] - x_list[-2]) < precision:  # Check for convergence
            break
    return x_list

# Number of iterations for gradient descent
n = 100

# Running gradient descent for different learning rates
low_gamma = gradient_descent(derivative_func=dg, initial_guess=3, multiplier=0.0005, precision=0.0001, max_iter=n)
mid_gamma = gradient_descent(derivative_func=dg, initial_guess=3, multiplier=0.001, precision=0.0001, max_iter=n)  # Mid learning rate
high_gamma = gradient_descent(derivative_func=dg, initial_guess=3, multiplier=0.002, precision=0.0001, max_iter=n)  # High learning rate

# Experiment with a very high learning rate (insane_gamma)
insane_gamma = gradient_descent(derivative_func=dg, initial_guess=1.9, multiplier=0.25, precision=0.0001, max_iter=n)

# Plotting the results for the different learning rates
plt.figure(figsize=(10, 6))

# Plot low_gamma results
plt.plot(low_gamma, label="Low gamma (0.0005)", color="blue")

# Plot mid_gamma results
plt.plot(mid_gamma, label="Mid gamma (0.001)", color="green")

# Plot high_gamma results
plt.plot(high_gamma, label="High gamma (0.002)", color="red")

# Plot insane_gamma results for experimentation
plt.plot(insane_gamma, label="Insane gamma (0.25)", color="orange", linestyle="--")

# Adding labels and title
plt.title("Gradient Descent with Different Learning Rates")
plt.xlabel("Iteration")
plt.ylabel("x Value")
plt.legend()
plt.grid(True)

# Display the plot
plt.show()
````

````python
## TASK-7 : Plot reduction in cost for each iteration
import numpy as np
import matplotlib.pyplot as plt

# Define the function g(x) = x^2 and its derivative dg(x)/dx = 2x
def g(x):
    return x**2  # Cost function (objective function)

def dg(x):
    return 2*x  # Derivative of the cost function

# Gradient Descent Function that also tracks cost reduction
def gradient_descent(derivative_func, initial_guess, multiplier, precision, max_iter):
    x = initial_guess
    x_list = [x]
    cost_list = [g(x)]  # Track the cost (f(x) = g(x))

    for _ in range(max_iter):
        gradient = derivative_func(x)
        x = x - multiplier * gradient  # Update x based on the gradient and multiplier
        x_list.append(x)
        cost_list.append(g(x))  # Track the cost at each iteration

        if abs(x_list[-1] - x_list[-2]) < precision:  # Check for convergence
            break

    return x_list, cost_list  # Return both x values and cost values

# Number of iterations for gradient descent
n = 100

# Running gradient descent for different learning rates (as before)
low_gamma_x, low_gamma_cost = gradient_descent(derivative_func=dg, initial_guess=3, multiplier=0.0005, precision=0.0001, max_iter=n)
mid_gamma_x, mid_gamma_cost = gradient_descent(derivative_func=dg, initial_guess=3, multiplier=0.001, precision=0.0001, max_iter=n)
high_gamma_x, high_gamma_cost = gradient_descent(derivative_func=dg, initial_guess=3, multiplier=0.002, precision=0.0001, max_iter=n)

# Plotting the cost reduction for each learning rate
plt.figure(figsize=(10, 6))

# Plot the cost reduction for low_gamma
plt.plot(low_gamma_cost, label="Low gamma (0.0005)", color="blue")

# Plot the cost reduction for mid_gamma
plt.plot(mid_gamma_cost, label="Mid gamma (0.001)", color="green")

# Plot the cost reduction for high_gamma
plt.plot(high_gamma_cost, label="High gamma (0.002)", color="red")

# Adding labels and title
plt.title("Cost Reduction for Each Iteration (Gradient Descent)")
plt.xlabel("Iteration")
plt.ylabel("Cost (f(x) = g(x))")
plt.legend()
plt.grid(True)

# Display the plot
plt.show()
````

#### Batch Gradient Descent with SymPy

````python
import numpy as np
from sympy import symbols, diff

# Define symbolic variables
a, b = symbols('x y')

# Define the cost function f(x, y) (example: f(x, y) = x^2 + y^2)
f = a**2 + b**2  # Example cost function

# Setup parameters
multiplier = 0.1
max_iter = 500
params = np.array([1.8, 1.0])  # initial guess

# Gradient Descent loop
for n in range(max_iter):
    # Compute the partial derivatives at the current parameters
    gradient_x = diff(f, a).evalf(subs={a: params[0], b: params[1]})
    gradient_y = diff(f, b).evalf(subs={a: params[0], b: params[1]})

    # Store gradients
    gradients = np.array([gradient_x, gradient_y])

    # Update parameters using gradient descent
    params = params - multiplier * gradients

# Results
print('Values in gradient array:', gradients)
print('Minimum occurs at x value of:', params[0])
print('Minimum occurs at y value of:', params[1])
print('The cost is:', f.evalf(subs={a: params[0], b: params[1]}))
````

#### Partial Derivatives of MSE w.r.t. $\theta_0$ and $\theta_1$

#### $$\frac{\partial MSE}{\partial \theta_0} = - \frac{2}{n} \sum_{i=1}^{n} \big( y^{(i)} - \theta_0 - \theta_1 x^{(i)} \big)$$

#### $$\frac{\partial MSE}{\partial \theta_1} = - \frac{2}{n} \sum_{i=1}^{n} \big( y^{(i)} - \theta_0 - \theta_1 x^{(i)} \big) \big( x^{(i)} \big)$$

#### MSE & Gradient Descent

````python
# x values, y values, array of theta parameters (theta0 at index 0 and theta1 at index 1)
def grad(x, y, thetas):
    n = y.size

    # Challenge: Create theta0_slope and theta1_slope to hold slope values from partial derivs
    theta0_slope = (-2/n) * sum(y - thetas[0] - thetas[1]*x)
    theta1_slope = (-2/n) * sum((y - thetas[0] - thetas[1]*x)*x)

    #return np.array([theta0_slope[0], theta1_slope[0]])
    #return np.append(arr=theta0_slope, values=theta1_slope)
    return np.concatenate((theta0_slope, theta1_slope), axis=0)
````

````python
multiplier = 0.01
thetas = np.array([2.9, 2.9])

# Collect data points for scatter plot
plot_vals = thetas.reshape(1, 2)
mse_vals = mse(y_5, thetas[0] + thetas[1]*x_5)

for i in range(1000):
    thetas = thetas - multiplier * grad(x_5, y_5, thetas)

    # Append the new values to our numpy arrays
    plot_vals = np.concatenate((plot_vals, thetas.reshape(1, 2)), axis=0)
    mse_vals = np.append(arr=mse_vals, values=mse(y_5, thetas[0] + thetas[1]*x_5))

# Results
print('Min occurs at Theta 0:', thetas[0])
print('Min occurs at Theta 1:', thetas[1])
print('MSE is:', mse(y_5, thetas[0] + thetas[1]*x_5))
````
