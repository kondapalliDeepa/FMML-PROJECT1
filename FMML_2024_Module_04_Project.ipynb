### Gradient Descent

````python
# Assuming df and f are defined elsewhere in your code
# Example definitions (you need to replace these with your actual functions):
# def df(x):
#     return 2 * x  # Example: derivative of f(x) = x^2

# def f(x):
#     return x**2  # Example: f(x) = x^2

# Gradient Descent
new_x = 3
previous_x = 0
step_multiplier = 0.1
precision = 0.00001

x_list = [new_x]
slope_list = [df(new_x)]

for n in range(500):
    previous_x = new_x
    gradient = df(previous_x)  # Calculate the gradient (derivative) at the current point
    new_x = previous_x - step_multiplier * gradient  # Update new_x using the gradient

    step_size = abs(new_x - previous_x)

    x_list.append(new_x)
    slope_list.append(df(new_x))  # Append the new slope (derivative) at the updated point

    if step_size < precision:
        print('Loop ran this many times:', n)
        break

print('Local minimum occurs at:', new_x)
print('Slope or df(x) value at this point is:', df(new_x))
print('f(x) value or cost at this point is:', f(new_x))
````

````python
def gradient_descent(derivative_func, initial_guess, multiplier=0.02, precision=0.001,
                    max_iter=300):
    new_x = initial_guess
    x_list = [new_x]
    slope_list = [derivative_func(new_x)]

    for n in range(max_iter):
        previous_x = new_x
        gradient = derivative_func(previous_x)
        new_x = previous_x - multiplier * gradient

        step_size = abs(new_x - previous_x)
        x_list.append(new_x)
        slope_list.append(derivative_func(new_x))

        if step_size < precision:
            break
    return new_x, x_list, slope_list
````

#### MSE & Gradient Descent

````python
# x values, y values, array of theta parameters (theta0 at index 0 and theta1 at index 1)
def grad(x, y, thetas):
    n = y.size

    # Challenge: Create theta0_slope and theta1_slope to hold slope values from partial derivs
    theta0_slope = (-2/n) * sum(y - thetas[0] - thetas[1]*x)
    theta1_slope = (-2/n) * sum((y - thetas[0] - thetas[1]*x)*x)

    #return np.array([theta0_slope[0], theta1_slope[0]])
    #return np.append(arr=theta0_slope, values=theta1_slope)
    return np.concatenate((theta0_slope, theta1_slope), axis=0)
````

````python
multiplier = 0.01
thetas = np.array([2.9, 2.9])

# Collect data points for scatter plot
plot_vals = thetas.reshape(1, 2)
mse_vals = mse(y_5, thetas[0] + thetas[1]*x_5)

for i in range(1000):
    thetas = thetas - multiplier * grad(x_5, y_5, thetas)

    # Append the new values to our numpy arrays
    plot_vals = np.concatenate((plot_vals, thetas.reshape(1, 2)), axis=0)
    mse_vals = np.append(arr=mse_vals, values=mse(y_5, thetas[0] + thetas[1]*x_5))

# Results
print('Min occurs at Theta 0:', thetas[0])
print('Min occurs at Theta 1:', thetas[1])
print('MSE is:', mse(y_5, thetas[0] + thetas[1]*x_5))
````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

# Example data (for demonstration purposes)
X = np.array([[1, 1], [1, 2], [2, 2], [2, 3]])  # Feature matrix
y = np.array([1, 2, 2, 3])  # Target values

# Function to compute Mean Squared Error
def compute_mse(theta0, theta1, X, y):
    predictions = theta0 + theta1 * X[:, 1]  # Linear model: y = theta0 + theta1 * x
    mse = np.mean((predictions - y) ** 2)  # Mean Squared Error
    return mse

# Generate theta values for plotting
theta0_range = np.linspace(-3, 3, 100)  # Range for theta0
theta1_range = np.linspace(-3, 3, 100)  # Range for theta1

# Create meshgrid for theta0 and theta1
plot_t0, plot_t1 = np.meshgrid(theta0_range, theta1_range)

# Compute the cost (MSE) for each combination of theta0 and theta1
plot_cost = np.zeros(plot_t0.shape)
for i in range(plot_t0.shape[0]):
    for j in range(plot_t0.shape[1]):
        plot_cost[i, j] = compute_mse(plot_t0[i, j], plot_t1[i, j], X, y)

# Collecting plot values and corresponding MSE values
plot_vals = np.column_stack((plot_t0.ravel(), plot_t1.ravel()))  # Shape (10000, 2)
mse_vals = plot_cost.ravel()  # Shape (10000,)

# Plotting MSE
fig = plt.figure(figsize=[16, 12])
ax = fig.add_subplot(111, projection='3d')  # Use add_subplot for 3D projection

ax.set_xlabel('Theta 0', fontsize=20)
ax.set_ylabel('Theta 1', fontsize=20)
ax.set_zlabel('Cost - MSE', fontsize=20)

# Scatter plot for the values of theta and their corresponding MSE
ax.scatter(plot_vals[:, 0], plot_vals[:, 1], mse_vals, s=80, color='black', label='MSE Points')

# Surface plot for the MSE
ax.plot_surface(plot_t0, plot_t1, plot_cost, cmap=cm.rainbow, alpha=0.4)

plt.legend()
plt.show()
````
